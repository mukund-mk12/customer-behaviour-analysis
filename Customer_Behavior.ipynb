```python
import pandas as pd
```

```python
cust = pd.read_csv("/content/customer_shopping_behavior.csv")
```

```python
cus = cust.copy()
```

```python
cus
```

```python
cus.isnull().sum()
```

```python
cus[cus["Review Rating"].isnull()]["Review Rating"]
```

```python
cus["Review Rating"] = cus.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

```python
cus.isnull().sum()
```

```python
cus.columns = cus.columns.str.lower()
```

```python
cus.columns = cus.columns.str.replace(" " ,"_")
```

```python
cus.head()
```

```python
cus["frequency_of_purchases"].value_counts()
```

```python
cus.info()
```

```python
# creating a age_group column
labels = ["Young Adults", "Adults", "Middle Aged", "Senior"]
cus["age_group"] = pd.qcut(cus["age"], q=4 , labels = labels)
```

```python
cus[["age","age_group"]]
```

```python
cus["frequency_of_purchases"].value_counts()
```

```python
# creating a Purchase_Frequency column
frequency_maping = {
"Every 3 Months" : 90,
"Annually"	: 365,
"Quarterly"	: 120,
"Monthly"	: 30,
"Bi-Weekly"	: 14,
"Fortnightly"	: 14,
"Weekly" : 7
}
```

```python
cus["purchase_frequency"] = cus["frequency_of_purchases"].map(frequency_maping)
```

```python
cus["purchase_frequency"]
```

```python
(cus["discount_applied"] == cus["promo_code_used"]).all()
```

```python
cus = cus.drop("promo_code_used",axis=1)
```
